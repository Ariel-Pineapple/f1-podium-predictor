## Paso 3.4 Construir el DataFrame

```python
dataset = results.merge(
    drivers[["driverId","surname"]],
    on="driverId"
)
```

```python
dataset = dataset.merge(
    constructors[["constructorId","name"]],
    on="constructorId"
)
```

```python
dataset = dataset.merge(
    races[["raceId","year","circuitId"]],
    on="raceId"
)
```

```python
dataset = dataset.rename(columns={
    "surname":"driver",
    "name_x":"constructor",
    "name_y":"circuit"
})
```

## Paso 3.6 Crear la variable objetivo

```python
dataset["podium"] = (
    dataset["positionOrder"] <= 3
).astype(int)
```

```python
dataset["podium"] = (
    dataset["positionOrder"] <= 3
).astype(int)
```

```python
# Renombrar constructor
dataset = dataset.rename(columns={
    "name": "constructor"
})
```

```python
dataset = dataset.merge(
    circuits[["circuitId", "circuitRef"]],
    on="circuitId"
)

dataset = dataset.rename(columns={
    "circuitRef": "circuit"
})
```
